````python
# X = [1,2,3]
#  x 1
#  x 2
#  x 3
def norm(X, p):
    # X : x1, x2 / x1, x2 ,x3
    total = 0 # variable
    for x in X: # x is element, X is list
        total = total + x ** p

    return total ** (1/p) # Hi value return kara, he function cha output ahe
    
````

````python
# L1 Norm of vector
norm([3,4], 1)
````

````python
# L2 Norm of vector
norm([3,4], 2)
````

````python
# As p increases, value of norm will go closer and closer to 4
# Therefore, L_max = max (3,4) = 4
norm([3,4], 15)
````

````python
# a is a vector
# a is a list (of numbers)
# a is array 
a = [6,8]
# L_max = 8
````

````python
norm(a,1)
````

````python
norm(a,2)
````
